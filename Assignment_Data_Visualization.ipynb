5. Generate a synthetic dataset with correlated features. Visualize the correlation matrix of a dataset using a
heatmap.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


np.random.seed(3)


x = np.random.rand(100)


y = 2 * x + np.random.normal(0, 0.1, 100)
z = -1.5 * x + np.random.normal(0, 0.1, 100)
w = x + y + np.random.normal(0, 0.2, 100)


data = np.vstack([x, y, z, w]).T


corr_matrix = np.corrcoef(data, rowvar=False)


plt.figure(figsize=(6, 5))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm',
            xticklabels=['X', 'Y', 'Z', 'W'],
            yticklabels=['X', 'Y', 'Z', 'W'])


plt.title('Correlation Matrix Heatmap')


plt.show()
```

3. Using the sales data, generate a heatmap to visualize the variation in sales across different months and

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


np.random.seed(20)
data = {
    'Month': np.random.choice(['Jan', 'Feb', 'Mar', 'Apr', 'May'], 100),
    'Day': np.random.choice(range(1, 31), 100),
    'Sales': np.random.randint(1000, 5000, 100)
}
df = pd.DataFrame(data)


sales_matrix = df.pivot_table(index='Day', columns='Month', values='Sales', aggfunc='mean')


plt.figure(figsize=(10, 6))
sns.heatmap(sales_matrix, annot=True, fmt=".0f", cmap='YlOrRd', cbar_kws={'label': 'Average Sales'})

plt.title('Heatmap of Average Sales Across Months and Days')
plt.xlabel('Month')
plt.ylabel('Day')

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))


data = {
    'X': x.flatten(),
    'Y': y.flatten(),
    'Z': z.flatten()
}
df = pd.DataFrame(data)


fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')


surf = ax.plot_surface(x, y, z, cmap='viridis', edgecolor='none', alpha=0.9)


fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10, label='Z value')


ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
ax.set_title('3D Surface Plot of z = sin(sqrt(x^2 + y^2))')


plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


np.random.seed(25)
data = {
    'Country': ['USA', 'Canada', 'UK', 'Germany', 'France'],
    'Population': np.random.randint(100, 1000, 5),
    'GDP': np.random.randint(500, 2000, 5)
}
df = pd.DataFrame(data)


plt.figure(figsize=(8,6))
plt.scatter(df['GDP'], df['Population'],
            s=df['Population']*2,
            alpha=0.6, c='teal', edgecolors='w')


for i, row in df.iterrows():
    plt.text(row['GDP']+5, row['Population']+5, row['Country'], fontsize=9)


plt.title('Bubble Chart: Population vs GDP')
plt.xlabel('GDP (in arbitrary units)')
plt.ylabel('Population (in millions)')

plt.grid(True)
plt.show()
```

3. Generate a Bokeh bar chart representing the counts of different fruits using the following dataset.
   fruits = ['Apples', 'Oranges', 'Bananas', 'Pears']
   counts = [20, 25, 30, 35]

```python
from bokeh.plotting import figure, show, output_notebook
from bokeh.models import ColumnDataSource


output_notebook()


fruits = ['Apples', 'Oranges', 'Bananas', 'Pears']
counts = [20, 25, 30, 35]


source = ColumnDataSource(data=dict(fruits=fruits, counts=counts))


p = figure(x_range=fruits, title="Fruit Counts",
           x_axis_label='Fruit', y_axis_label='Count',
           width=700, height=400)


p.vbar(x='fruits', top='counts', width=0.6, color='skyblue', source=source)


show(p)
```

5. Create a Bokeh heatmap using the provided dataset.
   data_heatmap = np.random.rand(10, 10)
 x = np.linspace(0, 1, 10)
 y = np.linspace(0, 1, 10)
 xx, yy = np.meshgrid(x, y)

```python
import numpy as np
from bokeh.plotting import figure, show, output_notebook
from bokeh.models import ColumnDataSource, LinearColorMapper, ColorBar
from bokeh.transform import transform
from bokeh.palettes import Viridis256


output_notebook()


data_heatmap = np.random.rand(10, 10)
x = np.linspace(0, 1, 10)
y = np.linspace(0, 1, 10)
xx, yy = np.meshgrid(x, y)

source = ColumnDataSource(data=dict(
    x=xx.flatten(),
    y=yy.flatten(),
    value=data_heatmap.flatten()
))


color_mapper = LinearColorMapper(palette=Viridis256, low=data_heatmap.min(), high=data_heatmap.max())

p = figure(title="Heatmap", x_axis_label='X', y_axis_label='Y',
           width=600, height=600, tools="hover", tooltips=[("Value", "@value")])


p.rect(x='x', y='y', width=(x[1]-x[0]), height=(y[1]-y[0]),
       source=source, fill_color=transform('value', color_mapper), line_color=None)


color_bar = ColorBar(color_mapper=color_mapper, label_standoff=12, location=(0,0))
p.add_layout(color_bar, 'right')


show(p)
```
